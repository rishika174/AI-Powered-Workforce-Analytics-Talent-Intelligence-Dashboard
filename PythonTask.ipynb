Create a table named students with workforce details and insert 20 employee records into it. Display all the records to verify the data has been inserted successfully.

```python
import pandas as pd
students = [
    {"StudentID": 1, "Name": "Alice", "Dept": "CSE", "Age": 20, "GPA": 8.5},
    {"StudentID": 2, "Name": "Bob", "Dept": "IT", "Age": 21, "GPA": 7.9},
    {"StudentID": 3, "Name": "Charlie", "Dept": "CSE", "Age": 19, "GPA": 9.2},
    {"StudentID": 4, "Name": "David", "Dept": "ECE", "Age": 22, "GPA": 8.1},
    {"StudentID": 5, "Name": "Eva", "Dept": "IT", "Age": 20, "GPA": 8.8},
    {"StudentID": 6, "Name": "Frank", "Dept": "CSE", "Age": 21, "GPA": 7.5},
    {"StudentID": 7, "Name": "Grace", "Dept": "ECE", "Age": 20, "GPA": 9.0},
    {"StudentID": 8, "Name": "Hank", "Dept": "IT", "Age": 19, "GPA": 8.4},
    {"StudentID": 9, "Name": "Ivy", "Dept": "CSE", "Age": 22, "GPA": 8.7},
    {"StudentID": 10, "Name": "Jack", "Dept": "ECE", "Age": 21, "GPA": 7.8}
]

df_students = pd.DataFrame(students)
print(df_students)
```

Write a Python statement to find the total number of students in the dataset.

```python
print("Total Students:", len(df_students))
```

Write code a solution to calculate the average GPA of all students, rounded to two decimal places.

```python

print("\nAverage GPA:", round(df_students["GPA"].mean(), 2))
```

Find the highest GPA among all students.

```python

print("\nHighest GPA:", df_students["GPA"].max())
```

Find the lowest GPA among all students.

```python

print("\nLowest GPA:", df_students["GPA"].min())
```

Calculate the average age of the students in years, rounded to two decimal places.

```python

print("\nAverage Age:", round(df_students["Age"].mean(), 2), "Years")
```

Display the total number of students in each department.

```python

print("\nDepartment-wise Student Count:")
print(df_students["Dept"].value_counts())
```

Filter and display the Student ID, Name, and Department of students with a GPA strictly greater than 8.5.

```python

print("\nStudents with GPA above 8.5:")
print(df_students[df_students["GPA"] > 8.5][["StudentID", "Name", "Dept"]])
```

Display the Student ID, Name, and GPA of all students belonging to the CSE department.

```python

print("\nStudents in CSE Department:")
print(df_students[df_students["Dept"] == "CSE"][["StudentID", "Name", "GPA"]])
```

Retrieve the top 5 students with the highest GPAs, showing their Student ID, Name, Department, and GPA.

```python

print("Top 5 Students by GPA:")
print(df_students.nlargest(5, "GPA")[["StudentID", "Name", "Dept", "GPA"]])
```

Find the total student count grouped by department.

```python

print("\nStudent Count by Department:")
print(df_students["Dept"].value_counts())
```

Calculate the average GPA for each department, rounded to two decimal places.

```python

print("\nAverage GPA by Department:")
print(df_students.groupby("Dept")["GPA"].mean().round(2))
```
